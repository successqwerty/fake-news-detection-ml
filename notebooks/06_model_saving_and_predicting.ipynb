```python
import pickle

with open("../model/logistic_model.pkl", "rb") as f:
    loaded_model = pickle.load(f)

with open("../model/tfidf_vectorizer.pkl", "rb") as f:
    loaded_vectorizer = pickle.load(f)
```

this function does -->  takes raw text , converts it using TF- idf , uses model to predict , returns readable output

```python
def predict_news(text):
    text_vectorized=loaded_vectorizer.transform([text])
    prediction=loaded_model.predict(text_vectorized)

    if prediction[0]==0:
        return "FAKE NEWS"
    else:
        return "REAL NEWS"
```

testing the model with the new text

```python
news_1 = "Government announces new education reform today"
news_2 = "Breaking shocking cure doctors don't want you to know"

print(predict_news(news_1))
print(predict_news(news_2))
```
